```python
import plotly.express as px

weekday_revenue = (
    daily_revenue_df
    .groupby("day_of_week", as_index=False)["revenue"]
    .mean()
)

fig = px.bar(
    weekday_revenue,
    x="day_of_week",
    y="revenue",
    title="Average Daily Revenue by Day of Week",
    labels={
        "day_of_week": "Day of week",
        "revenue": "Average revenue (€)"
    }
)

fig.show()
```
